# Create a CAS Connection using Swat

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from swat import *
cas = CAS(url,443, username, password)
cas.serverstatus()
```

# See available data

```python
cas.tableinfo(caslib='public')
```

```python
indata = cas.CASTable(name='HMEQ', caslib='public')
```

```python
indata.fetch(to=10)
```

```python
cas.loadActionSet('sampling')
cas.loadActionSet('decisionTree')
cas.loadActionSet('percentile')
```

# Define model inputs and partition data

```python
target = "BAD"   # binary target, expected event level "1"

inputs = [
    "LOAN", "MORTDUE", "VALUE", "YOJ",
    "DEROG", "DELINQ", "CLAGE", "NINQ", "CLNO", "DEBTINC",
    "JOB", "REASON"
]


# For a binary / nominal classification, include target + nominal predictors
nominals = ["BAD", "JOB", "REASON"]
```

```python
# ------------------------------------------------------------
# 6) Train / test split
#    Internal examples use sampling.srs with partind=True.
#    _PartInd_ = 1 -> train
#    _PartInd_ = 0 -> validation/test
# ------------------------------------------------------------
cas.sampling.srs(
    table=indata,
    samppct=70,
    seed=919,
    partind=True,
    output=dict(
        casOut=dict(name='indata_part', replace=True),
        copyVars='ALL'
    )
)
```

```python
indata_p = cas.CASTable(name='indata_part')
```

```python
indata_p.fetch(to=10)
```

```python
#make sure partition percentage is correct
train_fraction = indata_p['_PartInd_'].mean()
print(train_fraction)
```

# Train model

```python
rf_model = cas.decisionTree.forestTrain(
    table=dict(name=indata_p, where='_PartInd_ = 1'),
    target=target,
    inputs=inputs,
    nominals=nominals,
    nTree=500,
    maxLevel=10,
    leafSize=5,
    seed=12345,
    casOut=dict(name='rf_model', replace=True)
)

print("Forest model trained.")
print(rf_model)
```

# Score test set

```python
#8) Score the test partition
#    Internal examples use:
#      - model='rf_model'
#      - copyVars=target
#      - encodename=True
#      - assessonerow=True
# ------------------------------------------------------------
rf_score_obj = cas.decisionTree.forestScore(
    table=dict(name=indata_p, where='_PartInd_ = 0'),
    model='rf_model',
    casout=dict(name='rf_scored', replace=True),
    copyVars=target,
    encodename=True,
    assessonerow=True
)

print("\nScoring summary:")
print(rf_score_obj['ScoreInfo'])
```

# Evaluation

```python
score_col = f'P_{target}1'   # e.g. P_BAD1 or P_b_tgt1

rf_assess_obj = cas.percentile.assess(
    table='rf_scored',
    inputs=score_col,
    casout=dict(name='rf_assess', replace=True),
    response=target,
    event="1"
)

print("\nAssess output tables:")
print(rf_assess_obj['OutputCasTables'][['Name', 'Rows', 'Columns']])
```

```python

# ------------------------------------------------------------
# 10) Bring ROC / assessment table to the client
# ------------------------------------------------------------
roc_tbl = cas.CASTable(name='rf_assess_ROC').to_frame()

print("\nROC table preview:")
print(roc_tbl.head())
```

```python

# ------------------------------------------------------------
# 11) Accuracy / misclassification / confusion matrix at 0.5 cutoff
#    Internal examples use:
#      compare = df_assess[df_assess['_Cutoff_']==0.5]
#      misclassification = 1 - _ACC_
# ------------------------------------------------------------
cutoff_05 = roc_tbl[roc_tbl['_Cutoff_'] == 0.5].reset_index(drop=True)

if cutoff_05.empty:
    raise ValueError("No row found at cutoff 0.5 in rf_assess_ROC")

acc = float(cutoff_05.loc[0, '_ACC_'])
misclassification = 1.0 - acc

tp = int(cutoff_05.loc[0, '_TP_'])
fp = int(cutoff_05.loc[0, '_FP_'])
fn = int(cutoff_05.loc[0, '_FN_'])
tn = int(cutoff_05.loc[0, '_TN_'])

print("\nTest set performance at cutoff = 0.50")
print(f"Accuracy         : {acc:.4f}")
print(f"Misclassification: {misclassification:.4f}")
print(f"TP={tp}, FP={fp}, FN={fn}, TN={tn}")
```

```python

# ------------------------------------------------------------
# 12) AUC
#    The ROC table contains _C_ (AUC / concordance-style statistic)
#    and internal examples use it when plotting ROC legends.
# ------------------------------------------------------------
auc = float(roc_tbl['_C_'].mean())
print(f"AUC              : {auc:.4f}")
```

```python

# 13) Plot ROC curve
#    Internal examples plot:
#      x = _FPR_
#      y = _Sensitivity_
#      legend includes C statistic
# ------------------------------------------------------------
plt.figure(figsize=(8, 8))
plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Baseline')
plt.plot(
    roc_tbl['_FPR_'],
    roc_tbl['_Sensitivity_'],
    label=f'Random Forest (C={auc:.2f})'
)

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve - Random Forest')
plt.legend(loc='lower right')
plt.grid(True, alpha=0.3)
plt.show()
```
